## Review Python RegEx

Review the pages in the __[Python RegEx](https://www.w3schools.com/python/python_regex.asp)__ section.

#### Code Challenge

**Instructions**

Inside the editor, complete the following steps:

Import the re module

Create a variable txt with the value "The rain in Spain"

Search for "Spain" in txt and store the result in x

Print the position (span) of the match

```python
# Import re
import re

# Create a string
txt = "The rain in Spain"

# Search for "Spain"
x = re.search("Spain", txt)

# Print the span
print(x.span())
```
